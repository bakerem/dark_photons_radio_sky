## Power Spectrum Functions

Define functions to compute 2D power spectra from 3D lightcone boxes. The approach:
- `compute_2d_power`: Wraps powerbox to compute 2D power spectra (k_perp vs k_parallel)
- `powerspectra_2d`: Divides lightcone into chunks and computes power spectrum for each to improve statistics

This yields the dimensionless power spectrum as a function of perpendicular and parallel wavenumbers.

```python
def compute_2d_power(
   box,
   length,
   n_psbins,
   log_bins=True,
   ignore_kperp_zero=True,
):
    """
    Compute 2D power spectrum from a 3D box using powerbox.
    
    Wraps powerbox.get_power to Fourier transform and perform a cylindrical
    average in k-space (perpendicular vs parallel modes).
    
    Parameters
    ----------
    box : ndarray
        3D brightness temperature box
    length : tuple
        Physical dimensions (Mpc) in each direction
    n_psbins : int
        Number of bins for power spectrum
    log_bins : bool
        Use logarithmic binning
    ignore_kperp_zero : bool
        Exclude k_perp=0 mode from weighting
    
    Returns
    -------
    power, k_perp, k_parallel : tuple
        2D power spectrum and corresponding k-space bins
    """
    # Determine the weighting function required from ignoring k's
    k_weights = np.ones(box.shape[:2], dtype=int)
    n0 = k_weights.shape[0]
    n1 = k_weights.shape[-1]

    if ignore_kperp_zero:
        k_weights[n0 // 2, n0 // 2] = 0

    res = get_power(
        box,
        boxlength=length,
        bins=n_psbins,
        bin_ave=False,
        get_variance=False,
        log_bins=log_bins,
        k_weights=k_weights,
        res_ndim=2, 
    )

    res = list(res)
    k = res[1]
    if log_bins:
        k = np.exp((np.log(k[1:]) + np.log(k[:-1])) / 2)
    else:
        k = (k[1:] + k[:-1]) / 2

    res[1] = k
    return res

def powerspectra_2d(brightness_temp, n_psbins=25, nchunks=10, min_k=0.1, max_k=1.0, logk=True):
    """
    Compute 2D power spectra for multiple chunks of a lightcone.
    
    Divides the lightcone into chunks along the line-of-sight and computes
    the dimensionless power spectrum for each chunk to improve statistics.
    
    Parameters
    ----------
    brightness_temp : ndarray
        3D brightness temperature lightcone
    n_psbins : int
        Number of power spectrum bins
    nchunks : int
        Number of chunks to divide lightcone into
    logk : bool
        Use logarithmic k-space binning
    
    Returns
    -------
    data : list
        List of dicts containing k_perp, k_par, P, and delta for each chunk
    chunk_indices : list
        Indices marking chunk boundaries
    """
    data = []
    n_slices = roman.rs_array.shape[0]
    chunk_indices = list(range(0, n_slices, round(n_slices / nchunks)))

    if len(chunk_indices) > nchunks:
        chunk_indices = chunk_indices[:-1]
    chunk_indices.append(n_slices - 1)

    for i in tqdm(range(nchunks)):
        start = chunk_indices[i]
        end = chunk_indices[i + 1]
        cell_size = roman.box_len / roman.hii_dim
        chunklen = (end - start) * cell_size
        comoving_size = np.max(roman.lat) * roman.cosmo.comoving_distance(
            np.average([roman.rs_array[start], roman.rs_array[end-1]])
        ).value
        
        power, k_perp, k_parallel = compute_2d_power(
            brightness_temp[:, :, start:end],
            (comoving_size, comoving_size, chunklen),
            n_psbins,
            log_bins=logk,
        )
        
        # Compute magnitude of k vector and dimensionless power
        k_mag = np.sqrt(k_perp[:, None]**2 + k_parallel[0][None, :]**2)
        data.append({
            "k_perp": k_perp, 
            "k_par": k_parallel[0], 
            "P": power, 
            "delta": k_mag * power / (2 * np.pi**2)
        })
    return data, chunk_indices
```

## Compute 2D Power Spectra

Run the power spectrum calculation on the dark photon lightcone, dividing it into chunks for better statistics.

```python
powerspectra_2d_results, chunk_indices = powerspectra_2d(roman.dp_lightcone)
```

## Compute Foreground Wedge Parameters

Calculate the parameters needed to define the foreground wedge in k-space. The wedge slope depends on the Hubble parameter and angular size at each redshift chunk.

```python
import astropy.constants as const

# Calculate redshifts and wedge slopes for each chunk
chunk_zs = roman.rs_array[np.array(chunk_indices)]
average_zs = (chunk_zs[:-1] + chunk_zs[1:]) / 2

# Wedge slope depends on H(z) and comoving distance
slopes = (roman.cosmo.H(average_zs) * roman.cosmo.comoving_distance(average_zs) / 
          (const.c * (1 + average_zs))).to(un.dimensionless_unscaled)
```

## Visualize 2D Power Spectrum

Plot the dimensionless power spectrum in 2D k-space for a single redshift chunk. The shaded region shows the foreground wedge where astrophysical foregrounds dominate and obscure the cosmological signal. This represents an optimistic foreground removal scenario.

```python
# Select a chunk to visualize (chunk 8 at z ~ 5-6)
chunk = 8
print(f"Redshift: {average_zs[chunk]:.2f}")

# Extract k-space bins and power spectrum
k_perp = powerspectra_2d_results[chunk]["k_perp"]
k_par = powerspectra_2d_results[chunk]["k_par"]

# Define foreground wedge boundary (optimistic removal scenario)
wedge = 0.1 * roman.cosmo.H0.value / 100 + slopes[chunk] / 2 * k_perp

# Plot 2D power spectrum (only positive k_parallel)
plt.pcolormesh(
    k_perp, 
    k_par[k_par.shape[0]//2+1:], 
    np.swapaxes(powerspectra_2d_results[chunk]["delta"][:, k_par.shape[0]//2+1:], 0, 1), 
    norm="log",
    cmap="plasma",
)

# Shade the foreground wedge region
plt.fill_between(k_perp, wedge, color="black", alpha=0.35)

plt.xscale("log")
plt.yscale("log")

cbar = plt.colorbar()
cbar.set_label(r"$T_{\gamma,0}^2 \Delta^2_{\gamma \to A'}$ [mK$^2$]")
plt.xlabel(r"$k_{\perp}$ [Mpc$^{-1}$]")
plt.ylabel(r"$k_{\parallel}$ [Mpc$^{-1}$]")
plt.ylim(0.01, 2)
plt.xlim(0.025, 7.6)

plt.savefig("plots/21cm_ps.pdf", bbox_inches="tight")
plt.savefig("plots/21cm_ps.png", dpi=600, bbox_inches="tight")
```
